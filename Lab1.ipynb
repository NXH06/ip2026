# Setup for exercise B

```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt
```

```python
img_name = input("Image name: ")
read_img = cv.imread(f"{img_name}.jpg", cv.IMREAD_GRAYSCALE)
if read_img is None:
    raise FileNotFoundError(f"Image \"{img_name}.jpg\" not found.")
img: np.ndarray = read_img
read_img_color = cv.imread(f"{img_name} (color).jpg", cv.IMREAD_GRAYSCALE)
if read_img_color is None:
    raise FileNotFoundError(f"Image \"{img_name} (color).jpg\" not found.")
img_color: np.ndarray = read_img_color
```

# B1

## B1a

```python
pixels_original = img.ravel()

plt.figure(figsize=(8, 4))
plt.title("Histogram - Original Image")
plt.xlabel("Pixel intensity")
plt.ylabel("Frequency")
plt.hist(pixels_original, bins=256, range=(0, 255))
plt.xlim([0, 255])
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

img_equalized = cv.equalizeHist(img)
pixels_equalized = img_equalized.ravel()

plt.figure(figsize=(8, 4))
plt.title("Histogram - Equalized Image")
plt.xlabel("Pixel intensity")
plt.ylabel("Frequency")
plt.hist(pixels_equalized, bins=256, range=(0, 255))
plt.xlim([0, 255])
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## B1b

```python
img_resized = cv.resize(src=img, dsize=None, fx=1.5, fy=1.5, interpolation=cv.INTER_LINEAR)
```

```python
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(img, cmap="gray")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.title("Equalized")
plt.imshow(img_equalized, cmap="gray")
plt.axis("off")
plt.show()
```

## B1d

```python
def print_info(name: str, img: np.ndarray):
    height, width = img.shape[:2]
    channels = 1 if img.ndim == 2 else img.shape[2]
    dtype = img.dtype
    value_min = img.min()
    value_max = img.max()
    print(f"{name}:")
    print(f"\tSize (width x height): {width} x {height}")
    print(f"\tNumber of channels: {channels}")
    print(f"\tData type: {dtype}")
    print(f"\tMinimum pixel value: {value_min}")
    print(f"\tMaximum pixel value: {value_max}")
    print("-" * 50)

print_info("Color image", img_color)
print_info("Grayscale image", img)
```

# B2

```python

def adjust_brightness_and_contrast(image: np.ndarray, a: float, b: float):
    img_float = image.astype(np.float32) # prevent overflow for uint8 dtype in case of negative parameter(s) 
    output = a * img_float + b
    output = np.clip(output, 0, 255)
    return output.astype(np.uint8)

parameters = [(1, 50),
(1, -50),
(1.5, 0),
(0.5, 0)]
processed_images = []
histograms = []
for a, b in parameters:
    img_processed = adjust_brightness_and_contrast(img, a, b)
    processed_images.append((img_processed, a, b)) # store parameters with corresponding processed image for displaying info later
    histograms.append((img_processed.ravel(), a, b))

# Original and processed images
plt.figure(figsize=(12, 3))
plt.subplot(1, 5, 1)
plt.title("Original")
plt.imshow(img, cmap="gray")
plt.axis("off")
for idx, (img_proc, a, b) in enumerate(processed_images, start=2):
    plt.subplot(1, 5, idx)
    plt.title(f"a = {a}, b = {b}")
    plt.imshow(img_proc, cmap="gray")
    plt.axis("off")
plt.tight_layout()
plt.show()

# Histograms
plt.figure(figsize=(12, 3))
plt.subplot(1, 5, 1)
plt.title("Original")
plt.hist(img.ravel(), bins=256, range=(0, 255), color="steelblue")
plt.xlim([0, 255])
plt.grid(True, ls='--', alpha=0.5)
plt.ylim(0, 4000)
for idx, (pixels, a, b) in enumerate(histograms, start=2):
    plt.subplot(1, 5, idx)
    plt.title(f"a = {a}, b = {b}")
    plt.hist(pixels, bins=256, range=(0, 255), color="red")
    plt.xlim([0, 255])
    plt.grid(True, ls='--', alpha=0.5)
    plt.ylim(0, 4000)
plt.tight_layout()
plt.show()
```

| a | b | Effect |
|:-|:-|:-|
| 1 | 50 | brighter |
| 1 | -50 | darker |
| 1.5 | 0 | higher contrast |
| 0.5 | 0 | lower contrast |

# B3

```python
plt.figure(figsize=(12, 6))
plt.title("Original")
plt.hist(img.ravel(), bins=256, range=(0, 255), histtype="stepfilled")
plt.xlim([0, 255])
plt.grid(True, ls='--', alpha=0.5)

thresholds = [80, 120, 160]
processed_images = []
histograms = []
for t in thresholds:
    img_processed = cv.threshold(img, thresh=t, maxval=255, type=cv.THRESH_BINARY)
    img_processed = img_processed[1]
    processed_images.append((img_processed, t))
    histograms.append((img_processed.ravel(), t))

# Original and processed images
plt.figure(figsize=(12, 6))
plt.subplot(1, 4, 1)
plt.title("Original")
plt.imshow(img, cmap="gray")
plt.axis("off")
for idx, (img_proc, t) in enumerate(processed_images, start=2):
    plt.subplot(1, 4, idx)
    plt.title(f"Threshold = {t}")
    plt.imshow(img_proc, cmap="gray")
    plt.axis("off")
plt.tight_layout()
plt.show()
```
